```python
import numpy as np
```

```python
爻表 = {25: "xxxxxxxxx", 21: "---   ---",
        17: "---------", 13: "[=]   [=]"}

卦 = [np.zeros(3).astype(int) for 爻號 in range(6)]
爻號 = 0
數號 = 0

數 = 50
太極 = 1
數 = 數 - 太極
```

```python
比 = 0.71
assert 0<比<1
陽 = int(數 * 比)
人 = 1
陰 = 數 - 陽 - 人
陽閏 = (陽 - 1) % 4 + 1
陰閏 = (陰 - 1) % 4 + 1
卦[-爻號-1][數號] = 陽閏 + 人 + 陰閏
數 = 數 - 卦[-爻號-1][數號]

數號 = 數號 + 1
if 數號==3:
  卦[-爻號-1] = 爻表[np.sum(卦[-爻號-1])]
  爻號 = 爻號 + 1
  數號 = 0

  數 = 50
  太極 = 1
  數 = 數 - 太極

print(*卦, sep = "\n")
```
